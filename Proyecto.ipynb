```python
# ANALISIS EXPLORATORIO DE DATOS (AED) - NBA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configuración de visualización
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Cargar los datos
# Nota: Asegúrate de que el archivo CSV esté en la misma carpeta que este notebook
try:
    df = pd.read_csv('regular_season_totals_2010_2024.csv')
    print("Dataset cargado exitosamente.")
except FileNotFoundError:
    print("Error: El archivo CSV no se encontró. Verifica la ruta.")

# Mostrar información básica del dataset
print("="*80)
print("INFORMACIÓN GENERAL DEL DATASET")
print("="*80)
print(f"Dimensiones del dataset: {df.shape}")
print(f"\nPrimeras filas del dataset:")
display(df.head())

print("\n" + "="*80)
print("CLASIFICACIÓN DE VARIABLES")
print("="*80)

# Clasificar variables
variables_cualitativas = []
variables_cuantitativas_discretas = []
variables_cuantitativas_continuas = []

for col in df.columns:
    dtype = df[col].dtype
    if dtype == 'object' or df[col].nunique() < 10:
        variables_cualitativas.append(col)
    else:
        unique_vals = df[col].dropna().nunique()
        if pd.api.types.is_integer_dtype(dtype) and unique_vals < 50:
            variables_cuantitativas_discretas.append(col)
        else:
            variables_cuantitativas_continuas.append(col)

print(f"\nVariables Cualitativas ({len(variables_cualitativas)}):")
print(variables_cualitativas)

print(f"\nVariables Cuantitativas Discretas ({len(variables_cuantitativas_discretas)}):")
print(variables_cuantitativas_discretas[:10])

print(f"\nVariables Cuantitativas Continuas ({len(variables_cuantitativas_continuas)}):")
print(variables_cuantitativas_continuas[:10])

# Información detallada
info_df = pd.DataFrame({
    'Variable': df.columns,
    'Tipo': df.dtypes.values,
    'Valores Únicos': [df[col].nunique() for col in df.columns],
    'Valores Nulos': [df[col].isnull().sum() for col in df.columns],
    'Porcentaje Nulos': [f"{(df[col].isnull().sum()/len(df)*100):.2f}%" for col in df.columns]
})

display(info_df)
```

```python
# ANALISIS DESCRIPTIVO DE VARIABLES CUANTITATIVAS CONTINUAS 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

print("="*80)
print("ANÁLISIS DESCRIPTIVO - VARIABLES CUANTITATIVAS CONTINUAS")
print("="*80)

variables_analisis = [
    'MIN', 'FGM', 'FGA', 'FG_PCT', 'FG3M', 'FG3A', 'FG3_PCT',
    'FTM', 'FTA', 'FT_PCT', 'REB', 'AST', 'PTS', 'PLUS_MINUS'
]

estadisticas_df = pd.DataFrame(index=['Media', 'Mediana', 'Moda', 'Desv. Estándar', 
                                       'Varianza', 'Coef. Variación', 'Mínimo', 
                                       'Máximo', 'Rango', 'Q1', 'Q3', 'IQR'])

for var in variables_analisis:
    if var in df.columns:
        datos = df[var].dropna()
        media = np.mean(datos)
        mediana = np.median(datos)
        moda_result = stats.mode(datos, keepdims=True)
        moda = moda_result.mode[0]
        desv_std = np.std(datos, ddof=1)
        varianza = np.var(datos, ddof=1)
        coef_variacion = (desv_std / media) * 100 if media != 0 else np.nan
        minimo = np.min(datos)
        maximo = np.max(datos)
        rango = maximo - minimo
        q1 = np.percentile(datos, 25)
        q3 = np.percentile(datos, 75)
        iqr = q3 - q1
        
        estadisticas_df[var] = [
            f"{media:.4f}", f"{mediana:.4f}", f"{moda:.4f}",
            f"{desv_std:.4f}", f"{varianza:.4f}", f"{coef_variacion:.2f}%",
            f"{minimo:.4f}", f"{maximo:.4f}", f"{rango:.4f}",
            f"{q1:.4f}", f"{q3:.4f}", f"{iqr:.4f}"
        ]

display(estadisticas_df)
```
